# 1. 한눈에 보는 머신러닝

## 1.1 머신러닝이란

- 일반적 정의 : 머신러닝은 명시적 프로그래밍 없이 컴퓨터가 학습하는 능력을 갖추게 하는 연구 분야
- 공학적 정의 : let (작업 T : 경험 E $\to$ 성능 P $\uparrow\uparrow$) $\to$ ML : T와 P에 대해 E로 학습
    - ex) T = 스팸메일 분류 ; E = 훈련데이터 ; P = 직접정의(정확도 등)

- 기초 용어
    - 훈련 세트 : 시스템 학습 사용하는데 사용
    - 샘플(or 훈련 사례) : 훈련 세트의 각 데이터
    - 모델 : 학습 및 예측 수행 부분

## 1.2 왜 머신러닝을 사용하나요?

1) 기존 솔루션? $\to$ 많 (작업 + 유지보수) 필요 문제 : ML 모델이 코드 단순화, 더 잘 수행 (ex : 스팸필터)
2) 전통 솔루션? $\to$ 해결법 X인 복잡한 문제 : ML 기법 -> 해결책 O 일수도 (ex : 음성인식)
3) 유동적인 환경 : ML -> 새 데이터로 재훈련 OK. 적응 O (ex : 주식 예측)
4) 문제 + 데이터 -> 인사이트 얻기(데이터마이닝) : 숨겨진 패턴 -> 문제 이해 $\uparrow\uparrow$ (ex : 맥주와 기저귀간 상관관계)

## 1.3 애플리케이션 사례

- 제품 이미지 분석 -> 분류 : 이미지 분류 ; CNN, 비전 트랜스포머(TF)
- 뇌스캔 -> 종양 진단 : 시맨틱 분할 ; CNN, 비전 TF
- 자동 뉴스 기사 분류 : 자연어 처리(NLP)-텍스트 분류 ; RNN, CNN, TF
- 댓글 분류 : NLP
- 문서 요약 : NLP-텍스트 요약
- DNA 분석 및 개인 유전 위험도 예측 : SSM 모델
- 챗봇/개인 비서 : (NLP : 자연어 이해(NLU) + 질의응답 모듈)
- 기업 수익 예측 : 회귀 ; 회귀분석(선형/다항), SVM, 회귀 랜덤 포레스트, 회귀 NN, RNN,CNN,TF 등
- 음성 반응 앱 : 음성 인식 ; RNN,CNN,TF
- 부정거래 탐지 : 이상탐지 : 아이솔레이션 포레스트, 가우스 혼합 모델, AE 모델
- 고객 분류 : 군집 ; k-means, DBSCAN
- 고차원 데셋 -> 명확/의미 O 그래프 표현 : 데이터 시각화 ; 차원축소
- 고객별 맞춤형 추천 : 추천 시스템 ; NN 등
- 지능형 게임 봇 : AI 에이전트 ; RL 등

## 1.4 머신러닝 시스템의 종류

|**지도**|**회귀**|배치|사례 기반|전이|
|**자기 지도**|**분류**|온라인|모델 기반|앙상블|

### 1.4.1 훈련 안내 방식 (학습 패러다임) : 학습 동안 지도의 형태 or 정보량에 따라 분류

- 지도학습 : 훈련 데셋 -> 레이블(원하는 답) 포함
    - 분류 : 사례 -> 레이블 O?/X? ; 스팸 필터
    - 회귀 : 특성(feature) -> 타깃($\approx$ 레이블) 수치 예측 ;
        - 단, 분류 모델을 회귀에도 사용 O ex) 로지스틱 reg. : 클래스의 확률 출력

- 비지도학습 : 훈련 데셋 -> 레이블 X
    - 클러스터링(군집) 알고리즘 : 블로그 방문자 -> 특정 그룹으로 분류
        - 계층 군집 : 1차 분류된 군집을 더 작은 그룹으로 세분화 O
    - 시각화 : 대규모 고차원 데셋 -> 명확/의미 O인 2D/3D 그래프 만들어줌
        - 차원 축소 : $\approx$ 시각화 ; 정보손실 최소화 && 데이터 간소화.
            - 특성 추출 : 차원축소 방법 ; 여러 feat.을 통합해 적은 feat. 으로
    - 이상치 탐지 : 정상 샘플 훈련 -> (새 샘플? -> 정상?/이상치?(**outlier**))
        - 특이치 탐지 : 데이터 훈련 -> **기존 데이터 X**인 새 샘플? ($\therefore$ '깨끗한' 샘플 필요)
    - 연관 규칙 학습 : 특징간 흥미로운 상관관계 파악 ; BBQ소스+감자 -> 스테이크?

- 준지도 학습 : (훈련 데셋 : 레이블이 일부만 O) ; 지도+비지도로 구현
    - ex) (클러스터링 : 그룹 분류) -> (레이블 X? -> 그룹 레이블 부여) -> (if 레이블 OK?, then 지도학습)

- 자기 지도 학습 : (훈련 데셋 : 레이블 전혀 X) -> 레이블 완전 부여
    - ex) 고양이 이미지 맞추기? : (고양이 얼굴 가리기) -> (입력=가린 사진, 정답=원본 사진 -> 훈련)
    - 학습중 레이블 사용 $\therefore \approx$ 지도학습?
    - LLM 학습과 유사 : 말뭉치(corpus) -> 랜덤 단어 마스킹 -> 단어 예측
        - 전이학습 : 한 작업에서 다른 작업으로 지식 전달; DL(딥러닝)의 핵심 기술

- 강화학습 : (에이전트(학습하는 객체) : 환경 관찰 -> 행동 -> 보상/처벌 -> 정책(행동의 기준) 학습 -> 환경 관찰 ...)

### 1.4.2 배치 학습과 온라인 학습 : 입력 데이터의 스트림으로부터 점진적으로 학습 가능한가?

- 배치학습 : 모든 데이터를 사용해 훈련.
    - 오프라인 학습 : 더이상의 추가 학습 없이 (훈련 -> 적용) 과정
    - 데이터 드리프트 : 시간에 따라 모델 성능 $\downarrow \downarrow$

- 온라인 학습 : 미니배치라는 단위로 묶어 훈련. ex) 경사하강법
    - 언제? : 극도로 빠른 변화 적응 / 컴퓨팅 자원 제한 / (외부메모리 학습 : 매우 큰 데셋 분할)
    - 학습률 : 가장 중요 파라미터 ; 학습률 $\uparrow$? -> (적응 $\uparrow$ + 과거 데이터 망각 $\uparrow$) ; 낮으면 반대 
    - 단점 : 나쁜 데이터 -> 성능 감소

### 1.4.3 사례기반 학습과 모델 기반 학습 : 어떻게 일반화 되는가?

- 일반화 : 데이터로 학습 -> 새 데이터? good 예측 O

- 사례기반 학습 : 훈련 샘플을 기억해 학습 -> (새 샘플? -> 기존 샘플과 유사도 측정) -> 예측
- 모델 기반 학습 : 데이터를 설명할 수 있는 모델을 만들어 예측
    - 예제 : 2020의 삶 만족도(y)와 1인당 GDP(x)간 상관관계
        - 모델 선택 : 선형 모델 ; $y=\theta_0+\theta_1 x$ ($\theta$ : 모델 파라미터)
        - 모델이 얼마나 좋은가/나쁜가? : 효용함수(or 적합도 함수) vs 비용함수(or 손실함수)
        - 모델 훈련 : 데이터를 가장 잘 설명하는 (loss가 낮은) 파라미터의 값(계수)를 찾는 것

```python
# 모델 기반 학습
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.linear_model import LinearRegression

# 데이터 다운로드
data_root = 'https://github.com/ageron/data/raw/main'
lifesat_src = data_root + '/lifesat/lifesat.csv'
lifesat = pd.read_csv(lifesat_src)
X = lifesat[["GDP per capita (USD)"]].values
y = lifesat[["Life satisfaction"]].values

# 선형 모델 선택
model = LinearRegression()

# 모델 훈련
model.fit(X,y)

# 푸에트리코에 대한 예측
X_Puerto_Rico = [[33_442.8]] # 2020 푸에트리코 1인당 GDP
print('푸에트리코 예측 :', model.predict( X_Puerto_Rico))

# 데이터 그리기
lifesat.plot(kind='scatter', grid=True,
             x="GDP per capita (USD)", y="Life satisfaction")


plt.axis([23_500, 62_500, 4, 9])
reg_y1 = model.predict([[23500]])[0][0]
reg_y2 = model.predict([[62500]])[0][0]

print(reg_y1,reg_y2)
plt.plot([23500, 62500],[reg_y1,reg_y2], color='red')
plt.show()

```

```python
b0 = model.coef_
b1 = model.intercept_

print(b0,b1)

lifesat.plot(kind='scatter', grid=True,
             x="GDP per capita (USD)", y="Life satisfaction")
plt.axis([23_500, 62_500, 4, 9])
plt.show()
```

#### 더 많은 분류
- 앙상블 학습 : 여러 모델 훈련 -> 개별 예측 결합 -> 최종 예측 성능 향상
- 연합학습 : 탈중앙 방식 ; 모델이 여러 장치에서 훈련 -> raw 데이터 교환 없이 각 사용자에게 적용 ($\therefore$ 개인정보 보호)
- 메타학습 : 학습 방법을 학습 ; 최소한의 데이터로부터 새 작업을 빠르게 학습 O 방법?

## 1.5 머신러닝의 주요 도전과제

- 모델을 선택 -> 데이터를 훈련이 ML의 주요 과제 $\therefore$ '나쁜모델', '나쁜 데이터'가 문제가 된가

### 1.5.1 충분하지 않은 훈련 데이터

- 2009 논문 'The Unresonable Effectiveness of Data'(피터 노르빅 등) : 복잡한 문제에서 알고리즘보다 데이터가 더 중요하다
    - 그래도 여전히 소/중규모 데셋은 흔하고, 대규모 데셋을 구축하는건 쉽지 X. 아직은 알고리즘이 중요하긴 하다.

### 1.5.2 대표성 없는 훈련 데이터

- 좋은 일반화 : 훈련데이터가 일반화하고 싶은 사례를 잘 대표해야함
    - ex) GDP 예시에서, 23500 > or 62500 < 인 나라가 빠져서 대표성 완벽 X; -> 매우 가난하거나 부유한 국가는 잘못 예측할수도
- 샘플링 잡음 : 작은 표본크기에서, 샘플이 우연에 의한 대표성 없는 데이터로 구성, 모델 잘못 추청
- 잼플링 편향 : 표본이 크지만, 표본추출 방법 오류. 대표성 X

### 1.5.3 낮은 품질의 데이터
- 오류, 이상치, 잡음이 많은 샘플 -> 패턴 찾기 hard.
- $\therefore$ 데이터 정제 필수
    - 이상치 : 해당 샘플 무시 or 수동 수정
    - 빠진 feature : 해당 특성 무시 or 해당 샘플 무시 or 빠진값 대체(평균, 최빈값 ect.) or 따로 모델 훈련

### 1.5.4 관련 없는 특성
- GIGO : garbage in, garbage out
- 관련 X 특성이 적고, 관련 O 특성이 충분해야 학습 OK
- 특성공학(feature engineering) : 좋은 특성을 찾는 작업
    - feature selection : 가장 유용한 특성 선별
    - feat. extraction : 특성을 결합 -> 더 좋은 특성 합성 (차원축소 등을 통해)
    - 데이터 수집 : 새 데이터 수집해 새 특성 만듬

### 1.5.5 훈련 데이터 과대적합
- 모델이 훈련 데셋은 잘 예측하나, 새로운 케이스에는 잘 예측 X. 즉, 일반성 X
- 모델 복잡도에 비해 데셋이 작을경우, 잡음으로 인한 패턴을 학습
    - ex) GDP 예시 : w가 들어간 국가 (뉴질, 노르, 스웨, 스위) : 높 GDP, 높 만족도?
- regularization : 모델 단순, overfitting 회피
    - 자유도(모델에서 조절 가능한 변수의 갯수)를 줄이거나, 몇 변수를 고정하면 OK
    - 하이퍼 파라미터 : (모델이 아니라,) 학습 알고리즘의 파라미터. 훈련 전에 미리 지정되고, 훈련하는동안 상수로 남아있는 값.
        - 릿지/라쏘 회귀에서 규제에 대한 상수항을 높? -> 매우 강한 수준으로 규제

### 1.5.6 훈련 데이터 과소적합
- 과대적합과 반대로 모델이 너무 단순해서 데이터의 패턴을 학습 X
    - 모델 param.이 더 많은 모델
    - 더 좋은 feat. 추가
    - 모델 규제 완화

### 1.5.7 배포 문제
- 잘 정제 && 학습이 잘 되었어도 모델 배포에서 문제가 O
    - 복잡한 유지보수
    - 최적화 X; 메모리 너무 많이 잡아먹음
    - 너무 느림
    - 확장성 X
    - 보안 취약
    - 구식 업데이터

- MLOps : 이러한 ML에서 배포/인프라 문제 해결 직무

### 1.5.8 핵심 요약

- ML : 명시적 규칙 코딩 X; 데이터로부터 학습해 특정 작업을 개선
- ML 시스템 종류 : 지도/비지도, 배치/온라인, 사례기반/모델기반
- ML 프로젝트 과정 : 데셋 수집 -> 학습 알고 주입 -> 
    - (모델기반이면,) 모델 param 조정; 훈련세트에 적합한 예측 -> 새 데셋에서도 일반화 되는지?
    - (사례기반) 샘플 기억(=비지도 학습) -> 유사도측정 -> 새 샘플 vs 기존 샘플 비교해 예측
- ML 주요 이슈 : 작은 데셋, 대표성 X, 잡읍 많 & 관X 특성들, 너무 단순한/복잡한 모델, 배포/유지보수 측면도 고려

## 1.6 테스트와 검증

- 훈련 데셋을 훈련 세트와 테스트 세트로 나누어서 모델이 일반화가 잘 되었는지 평가
- 일반화 오치(or 외부 샘플 오차) : 새 샘플에 대한 오차 비율 (너무 크다? -> OF일 가능성도)

### 1.6.1 하이퍼파라미터 튜닝과 모델 선택

- 홀드아웃 검증 : 일반화 오차를 단일 테스트셋에서 여러번검증시 문제 -> 테스트 셋과는 별개로, 훈련 데셋에서 또 데이터를 떼와서 모델을 검증용으로만 사용 -> 가장 우수한 모델을 선별 및 훈련
    - 검증 세트(or 개발세트, 데브세트) : 이러한 홀드아웃 검증에 사용되는 세트
    - 과정
        1) 훈련세트 : 여러 모델 훈련
        2) 검증세트 : 모델 평가
        3) 훈련+검증 : 2)에서 가장 우수한 모델 재훈련
        4) 테스트셋 : 최종 모델 평가
- 교차검증 : 훈셋이 너무 작거나, 검셋이 너무 비대할 경우 적절 X ; $\therefore$ 작은 검증셋 여러개를 써서 검증. 검셋마다 나머지 데이터에서 훈련한 모델을 해당 검셋에서 평가. -> 모든 모델 평가를 평균내면 더 정확한 성능 측정 O

### 1.6.2 데이터 불일치
- 훈련 데이터가 실제 제품에 사용될 데이터를 완벽하게 대표하지 못할수도 O
    - ex) 사진 분류 모델에 쓰일 웹 이미지들 -> 모바일 카메라 사진 환경과 동일?
    - 검증세트와 테스트셋이 실전의 데이터를 잘 반영해야한다
    - 훈련-개발 세트 : 훈련 세트를 일부 떼어 또 다른 세트;
        - 웹 이미지 : 훈련 세트 + 훈련-개발세트
        - 모바일 이미지 : 검증세트 + 테스트세트

- (참고) 공짜점심없음(NFL) 이론 : 모델 선택시 이 모델이 데이터를 잘 예측할거란 가정을 포함.
    - 데이터에 아무런 가정이 없으면 한 모델이 다른 모델보다 선호될 이유가 없음
    - 특정 데셋엔 선형회귀가 맞지만 특정 데셋엔 신경망이 적합
    - 즉, 경험하기 전에 더 잘 맞을것이라 보장하는 모델은 없음. 선택이 옳았는지는 모델 평가에서만 OK
